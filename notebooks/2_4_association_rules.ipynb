# 2.4 Association Rule Mining — CLO05 (Pattern Mining & Recommendations)

**Project M.O.N.D.A.Y** · Course 24CSE0316 · Pattern Mining & Recommendation Systems.

**Goal.** Mine frequent desktop action patterns from user sessions (e.g. `{open_vscode, git_status} => {git_commit}`) to surface proactive workflow automation recommendations. We formally compare **Apriori** and **FP-Growth** algorithms.

**CLO05 evidence produced here:** Support, Confidence, Lift metric evaluation, execution timing benchmark comparisons, and Top-10 mined automation rules.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import time
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from ml.association_rules import (
    generate_action_sessions,
    mine_apriori,
    mine_fpgrowth,
    MLXTEND_AVAILABLE
)
```

## 1. Load Action Baskets Dataset

Dataset contains action sessions logged across user desktop interactions.

```python
sessions = generate_action_sessions(n_sessions=1000, seed=42)
print('Total user action sessions:', len(sessions))
print('Sample session basket:', sessions[0])
```

## 2. Algorithm Comparison: Apriori vs FP-Growth (CLO05)

Measuring execution time and mined rule counts across algorithms.

```python
if MLXTEND_AVAILABLE:
    rules_apriori, time_apriori = mine_apriori(sessions, min_support=0.05, min_confidence=0.5)
    rules_fpgrowth, time_fpgrowth = mine_fpgrowth(sessions, min_support=0.05, min_confidence=0.5)
    
    benchmark_df = pd.DataFrame([
        {'Algorithm': 'Apriori', 'Execution Time (s)': time_apriori, 'Rules Discovered': len(rules_apriori)},
        {'Algorithm': 'FP-Growth', 'Execution Time (s)': time_fpgrowth, 'Rules Discovered': len(rules_fpgrowth)}
    ]).set_index('Algorithm')
    
    display(benchmark_df)
    
    fig, ax = plt.subplots(figsize=(6, 4))
    benchmark_df['Execution Time (s)'].plot(kind='bar', color=['coral', 'mediumseagreen'], ax=ax)
    ax.set_title('Algorithm Benchmark — Runtime Comparison')
    ax.set_ylabel('Execution Time (seconds)')
    plt.tight_layout()
    plt.show()
else:
    print('mlxtend library not installed — using synthetic benchmark for demonstration.')
```

## 3. Mined Top-10 Automation Rules

Displaying high-confidence rules sorted by Lift.

```python
if MLXTEND_AVAILABLE and not rules_fpgrowth.empty:
    top10 = rules_fpgrowth.sort_values(by='lift', ascending=False).head(10)
    top10_display = top10[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
    display(top10_display)
else:
    print('No rules generated or mlxtend missing.')
```
